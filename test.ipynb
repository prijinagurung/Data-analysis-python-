```python
#READ THE CSV FILE

import pandas as pd #the library we will be using 
df = pd.read_csv("/Users/prijinagurung/Desktop/PythonA/Data-analysis-python-/pandas_dataset.csv") #importing the dataset
df
```

```python
df.head() #printing the first 5 rows
```

```python
#RENAME COLUMNS

#the updated dataset already has column names in lowercase and separated by _
```

```python
#DROP UNNECESSARY ROWS

df_dropped = df.drop(df.index[0], axis = 0) #removing the 1st row after header since it is blank
#axis = 0 shows that the operation is on a row 
df_dropped.head()

df_dropped = df_dropped.reset_index(drop = True) #resetting the index to 0,1,2,... from 1,2,3,...
#here, drop = True indicates that the old index will be deleted
df_dropped.head()
```

```python
#DROP COLUMNS

df_dropped = df_dropped.drop("order_date", axis = 1) 
#axis = 1 shows that the operation is on a column
df_dropped.head()

#I decided to drop the order_date column since there are no questions that require it for this particular assignment.
```

```python
#COUNT MISSING VALUES
print(df_dropped.isnull().sum()) #this displays the null values in each of the columns
```

```python
#HANDLE MISSING VALUES
#a) Drop rows based on column

df_dropped = df_dropped.dropna(subset = ['sales'], axis = 0)
#dropping rows (axis=0) using dropna having null values in sales column
df_dropped
```

```python
#HANDLE MISSING VALUES
#b) Fill missing values

df_dropped['profit'] = df_dropped['profit'].fillna(df_dropped['profit'].mean())
#fillna fills null values in profit column with the mean of existing values in profit

df_dropped['discount'] = df_dropped['discount'].fillna(df_dropped['discount'].median())
#filling null values in discount columnn with median

df_dropped
```

```python
#HANDLE MISSING CATEGORICAL DATA

df_dropped['customer_name'] = df_dropped['customer_name'].fillna('Unknown')
#giving a fixed string to replaces the null values with it rather than mean/median
df_dropped 
```

```python
#DETECT AND REMOVE DUPLICATES

duplicates = df_dropped.duplicated(subset = 'order_id').sum() #counting all the duplicate rows based on order id
print (f"We will remove {duplicates} duplicate rows")

df_dropped[df_dropped.duplicated(subset ='order_id')]
#checking for duplicated rows
```

```python
df_nodup = df_dropped.drop_duplicates(subset = ['order_id'])
df_nodup
```

```python
#FILTERING AND CREATING NEW COLUMNS

df_nodup[df_nodup['unit_price']>20000] #this displays the output based on the condition applied to dataframe 
```

```python
df_nodup[(df_nodup['unit_price']>10000) & (df_nodup['status']== "Completed")][['customer_name', 'category', 'status']]
#the first big brackets provides the condition that needs to be fulfilled to print
#the second big brackets gives the columns to print
```

```python
df_nodup['total_amount'] = df_nodup['quantity']*df_nodup['unit_price']-df_nodup['discount']
#creating a new column based on the calculation on existing columns
df_nodup
```

```python
df_nodup ['customer_type'] = df_nodup['quantity'].apply(lambda x: 'Bulk Buyer' if x>=3 else 'Regular Buyer')
#creating a new columun named customer_type based on the condition applied to quantityt column
#using lambda funciton with the condition makes it applicable to the entire quantity column
df_nodup
```
